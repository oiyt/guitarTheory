```python
import subprocess

def rip_url_to_mp3(media_url, output_mp3_path):
    # Build the ffmpeg command arguments
    command = [
        "ffmpeg",
        "-i", media_url,
        "-vn",            # Drop video track
        "-ar", "44100",   # Set audio sample rate to 44100 Hz
        "-ac", "2",       # Set to 2 audio channels (stereo)
        "-b:a", "192k",   # Set audio bitrate to 192 kbps
        output_mp3_path
    ]
    
    # Run the command in the background
    try:
        subprocess.run(command, check=True)
        print(f"Successfully saved to {output_mp3_path}")
    except subprocess.CalledProcessError as e:
        print(f"Error while ripping audio: {e}")

# Example usage:
url = "https://www.youtube.com/watch?v=pTdihu-mp90&time_continue=54&source_ve_path=NzY3NTg&embeds_referring_euri=https%3A%2F%2Fwww.google.com%2F"
output_file = "output_audio.mp3"

rip_url_to_mp3(url, output_file)
```
